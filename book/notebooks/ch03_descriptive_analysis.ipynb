## Code 3.1 (p. 53)

Script to generate the frequency table and plot a pie chart for variable “Shape” of the mammographic dataset.

**Book version**

```python
# Determining the frequency distribution, frequency table and pie chart
# of variable 'Shape' in the Mammographic dataset

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo

# fetch dataset (https://archive.ics.uci.edu/dataset/161/mammographic+mass)
dmammo = fetch_ucirepo(id=161)["data"]["original"]

SShape = pd.Series(dmammo['Shape'])
ftable = SShape.value_counts(dropna=False) # Generate the frequency table
rftable = ftable / len(SShape) * 100  # Relative frequency
cftable = ftable.cumsum() / len(SShape) * 100  # Cumulative frequency
df = pd.DataFrame({
    "Frequency": ftable.to_list(), 
    "Relative Frequency": rftable.to_list(), 
    "Cumulative Frequency": cftable.to_list()})

placeholders = {1.0:'Round', 2.0:'Oval', 3.0:'Lobular', 4.0:'Irregular'}
ftable.index = ftable.index.map(lambda x: placeholders.get(x, '?'))

print(df)
fig, figftable = plt.subplots()

# Using a color palette with different levels of the same color
colors = sns.color_palette("Blues", len(ftable)* 3)[::-3]

# Plotting the pie chart with the new color palette
figftable.pie(
    ftable.to_list(), labels=ftable.index.to_list(), 
    autopct='%1.2f%%', colors=colors)

# Show the plot
plt.show()
```

## Code 3.2 (p. 55)

Script to generate the frequency table and plot a histogram for variable “temp” of the Forest Fires dataset.

> **Corrected ([E12](https://lndecastro.github.io/EDA-Book/errata.html#e12)).** Bins are sorted before the cumulative frequency is computed (`value_counts()` sorts by frequency).

**Book version**

```python
# Determining the frequency distribution, frequency table and histogram
# of continuous variables in the Forest Fires dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo

# fetch dataset (https://archive.ics.uci.edu/ml/datasets/forest+fires)
dforest = fetch_ucirepo(id=162)["data"]["original"]

var = "temp" # Choose the target variable
SShape = pd.Series(dforest[var]) 
nbins = 10; inflimit = 0; suplimit = max(SShape)
ampl = (suplimit - inflimit) / nbins

# Define the range of the variable and bin size
fbins = np.arange(0, suplimit + ampl, ampl)

# The pandas.cut function groups data into bins and counts frequency
ftable = pd.cut(SShape, fbins).value_counts().sort_index()  # Absolute frequency
rftable = ftable / len(SShape) * 100  # Relative frequency
cftable = ftable.cumsum() / len(SShape) * 100  # Cumulative frequency
df = pd.DataFrame({"Bins": ftable.index.to_list(),
                   "Frequency": ftable.to_list(),
                   "Relative Frequency": rftable.to_list(),
                   "Cumulative Frequency": cftable.to_list(),})
print(df)
plt.xticks(fbins)
sns.histplot(dforest, x=var, bins=fbins, kde=True)

# Show the plot
plt.show()
```

## Code 3.7 (p. 65)

(a) Scripts to calculate the mean of a frequency distribution and (b) the weighted, geometric, harmonic, and trimmed means of a variable.

### Code 3.7a

**Book version**

```python
# Calculating the mean of a frequency distribution (Eq. 3.4)
# Create a DataFrame from the frequency distribution data

import pandas as pd

data = {
    "Bins": ["(16.65, 19.98]", "(19.98, 23.31]", "(13.32, 16.65]",
             "(23.31, 26.64]", "(9.99, 13.32]", "(26.64, 29.97]",
             "(3.33, 6.66]", "(6.66, 9.99]", "(29.97, 33.3]",
             "(0.0, 3.33]"],
    "Frequency": [128, 119, 75, 69, 47, 30, 20, 15, 13, 1]}
df = pd.DataFrame(data)

# Calculate the midpoint of each bin (interval)
df["Midpoint"] = (
    # Get first float from interval 
    df["Bins"].str.split(", ").str[0].str.replace("(", "").astype(float) +  
    # Get second float from interval
    df["Bins"].str.split(", ").str[1].str.replace("]", "").astype(float)
) / 2
# Multiply the midpoint by the frequency to get the product
df["Product_fx"] = df["Midpoint"] * df["Frequency"]

# Sum the products and frequencies
sprod = df["Product_fx"].sum()
sfreq = df["Frequency"].sum()

# Calculate the mean
mean = sprod / sfreq
print(df)
print(f"Mean of the frequency distribution: {mean:.2f}")
```

### Code 3.7b

> **Corrected ([E17](https://lndecastro.github.io/EDA-Book/errata.html#e17)).** Non-negative weights with a positive total are used: burned area, as in Code 3.8, gives an area-weighted summary. Geometric and harmonic means are not applied to "temp" (°C is an interval scale), so "wind" replaces it.

**Book version**

```python
# Calculate the weighted average (Eq. 3.5), geometric (Eq. 3.6)
# harmonic (Eq. 3.7), and trimmed (Eq. 3.8) means

import numpy as np
import scipy.stats as spy

variables = ["FFMC", "wind"]  # [E17] positive, ratio-scale variables
for var in variables:
    weights = dforest["area"]  # [E17] non-negative weights (area-weighted summary)
    wavg = np.average(dforest[var], weights=weights)
    gavg = spy.gmean(dforest[var])  # From Scipy library
    havg = spy.hmean(dforest[var])  # From Scipy library
    tavg = spy.trim_mean(dforest[var], 0.05)  # 5% trim

    print(f"Weighted average of variable {var}: {wavg:.2f}")
    print(f"Geometric mean of variable {var}: {gavg:.2f}")
    print(f"Harmonic mean of variable {var}: {havg:.2f}")
    print(f"Trimmed mean of variable {var}: {tavg:.2f}\n")
```

## Code 3.8 (p. 66)

Script to generate the central tendency measures for some variables of the Forest Fires dataset. The weights in the weighted mean were taken as the values in the “area” variable.

> **Corrected ([E19](https://lndecastro.github.io/EDA-Book/errata.html#e19)).** Geometric and harmonic means are omitted (`nan`) for temp, which is in °C (interval scale), and for ISI and rain, which contain zeros.

**Book version**

```python
# Central Tendency Measures for the Forest Fires Dataset
# Columns of interest: 'FFMC','DMC','DC', 'ISI', 'temp', 'RH', 'wind', 'rain'

import numpy as np
import scipy.stats as spy
from ucimlrepo import fetch_ucirepo

# fetch dataset 2
# https://archive.ics.uci.edu/ml/datasets/forest+fires
dforest = fetch_ucirepo(id=162)["data"]["original"]

# Dictionary to store the results
CTM = {}

variables = ["FFMC", "DMC", "DC", "ISI", "temp", "RH", "wind", "rain"]
for var in variables:
    # [E19] GM and HM omitted for temp (°C, interval scale) and zero-containing variables
    valid = bool((dforest[var] > 0).all()) and var != "temp"
    # Add the results to the dictionary
    CTM[var] = {
        "Mean": np.mean(dforest[var]),
        "Median": np.median(dforest[var]),
        "Midpoint": (np.max(dforest[var]) + np.min(dforest[var])) / 2,
        "Weighted Mean": np.average(dforest[var], weights=dforest["area"]),
        "Geometric Mean": spy.gmean(dforest[var]) if valid else np.nan,
        "Harmonic Mean": spy.hmean(dforest[var]) if valid else np.nan,
        "Trimmed Mean": spy.trim_mean(dforest[var], proportiontocut=0.1),
    }

# Print the results
for key, value in CTM.items():
    print(key)
    for stat_name, stat_value in value.items():
        print(f"\t{stat_name}: {stat_value:.2f}")
```

## Code 3.9 (p. 72)

Script to calculate the range, IQR, sIQR, variance, standard deviation, and CV of numeric variable “FFMC” of the Forest Fires dataset.

> **Corrected ([E24](https://lndecastro.github.io/EDA-Book/errata.html#e24), [E23](https://lndecastro.github.io/EDA-Book/errata.html#e23)).** Sample variance and standard deviation (`ddof=1`) to match Eqs. (3.12) and (3.14); the CV is printed as a percentage (6.09% is low relative variability).

**Book version**

```python
# Calculate the variability measures range (Eq. 3.9),
# IQR (Eq. 3.10), sIQR (Eq. 3.11), variance (Eq. 3.12),
# std (Eq. 3.14), and CV (Eq. 3.16) using Numpy

import numpy as np
from ucimlrepo import fetch_ucirepo

# fetch dataset
# https://archive.ics.uci.edu/ml/datasets/forest+fires
dforest = fetch_ucirepo(id=162)["data"]["original"]

var = "FFMC"
drange = np.max(dforest[var]) - np.min(dforest[var])
Q1, Q3 = np.percentile(dforest[var], [25, 75])
IQR = Q3 - Q1
sIQR = IQR / 2
dvar = np.var(dforest[var], ddof=1)  # [E24] sample variance, Eq. (3.12)
dstd = np.std(dforest[var], ddof=1)  # [E24] sample std, Eq. (3.14)
CV = dstd / np.mean(dforest[var]) * 100

print("*Variability Measures*")
print(f"Range of variable FFMC: {drange:.2f}")
print(f"IQR of variable FFMC: {IQR:.2f}")
print(f"sIQR of variable FFMC: {sIQR:.2f}")
print(f"Variance of variable FFMC: {dvar:.2f}")
print(f"Standard deviation of variable FFMC: {dstd:.2f}")
print(f"Coefficient of variation of variable FFMC: {CV:.2f}%")
```

## Code 3.11 (p. 77)

Scripts to generate skewed distributions using a beta distribution from Numpy. (a) Right-skewed distribution. (b) Left-skewed distribution.

### Code 3.11a

> **Corrected ([E32](https://lndecastro.github.io/EDA-Book/errata.html#e32)).** The midpoint is not the mode. The mode is now estimated as the peak of a kernel density estimate, and the Pearson first coefficient (mean − mode)/s is positive for the right-skewed illustrative sample and negative for the left-skewed one. A fixed seed is set.

> **Names corrected ([E29](https://lndecastro.github.io/EDA-Book/errata.html#e29))** in comments and labels.

**Book version**

```python
# Skewness and Skewed distributions
# Generate random data with a right-skewed distribution

import statistics as st
import numpy as np
from scipy.stats import skew, gaussian_kde
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)  # fixed seed for reproducibility
data = np.random.beta(a=1, b=5, size=1000) # Beta distribution
mean = st.mean(data)
median = st.median(data)
# [E32] Estimate the mode as the peak of a kernel density estimate
grid = np.linspace(min(data), max(data), 1000)
mode = grid[np.argmax(gaussian_kde(data)(grid))]
print(f"Mean, median, and mode: {mean:.2f} {median:.2f} {mode:.2f}")
print(f"Skewness (Fisher-Pearson Coefficient): {skew(data):.2f}")
print(f"Skewness (Pearson First Coefficient): {(mean - mode) / np.std(data, ddof=1):.2f}")

ax = sns.histplot(data, bins="auto", kde=2)
plt.axvline(x=mean, color="r", linestyle="--", label="Mean")
plt.axvline(x=median, color="g", linestyle="-", label="Median")
plt.axvline(x=mode, color="b", linestyle=":", label="Mode")
plt.legend()

# Show the plot
plt.show()
```

## Code 3.12 (p. 80)

Script to generate the four data distributions presented in Figure 3.6 to illustrate the different shapes and calculus of kurtosis.

> **Corrected ([E37](https://lndecastro.github.io/EDA-Book/errata.html#e37)).** A fixed random seed is set, so the kurtosis values are the same on every run.

**Book version**

```python
# Kurtosis: mesokurtic, platykurtic and leptokurtic distributions

import numpy as np
from scipy.stats import norm, laplace, semicircular, kurtosis
import seaborn as sns
import matplotlib.pyplot as plt

# Fixed random seed so that results are reproducible
np.random.seed(42)

# Set up sub-plots layout
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(13, 13), layout="constrained")

# Normal distribution (Mesokurtic)
dnorm = norm.rvs(size=10000)
knorm = kurtosis(dnorm)
print(f"Kurtosis: {knorm:.2f}")
sns.histplot(dnorm, bins="auto", kde=2, ax=ax1)
ax1.set_title(f"Normal Distribution - Kurtosis: {knorm:.2f}")

# Uniform distribution (Platykurtic)
dunif = np.random.uniform(0.01, 0.10, 10000)
kunif = kurtosis(dunif)
print(f"Kurtosis: {kunif:.2f}")
sns.histplot(dunif, bins="auto", kde=2, ax=ax2)
ax2.set_title(f"Uniform Distribution - Kurtosis: {kunif:.2f}")

# Laplace distribution (Leptokurtic)
dlap = laplace.rvs(loc=0, scale=1, size=10000)
klap = kurtosis(dlap)
sns.histplot(dlap, bins="auto", kde=2, ax=ax3)
ax3.set_title(f"Laplace Distribution - Kurtosis: {klap:.2f}")

# Wigner semicircle distribution
dwigner = semicircular.rvs(size=10000)
kwigner = kurtosis(dwigner)
sns.histplot(dwigner, bins="auto", kde=2, ax=ax4)
ax4.set_title(f"Wigner Semicircle Distribution - Kurtosis: {kwigner:.2f}")

# Show the plot
plt.show()
```

> **Note ([E38](https://lndecastro.github.io/EDA-Book/errata.html#e38), [E39](https://lndecastro.github.io/EDA-Book/errata.html#e39)).** The script uses the pandas methods `skew()` and `kurtosis()`, which return the bias-corrected sample skewness and excess kurtosis (normal = 0). Any positive value is labeled "Leptokurtic", so the output for wind reads "wind Kurtosis: 0.05 (Leptokurtic)"; values this close to zero indicate tails similar to those of a normal distribution.

**Book version**

```python
# Calculate the Skewness and Kurtosis for the Forest Fires variables

import pandas as pd
from scipy.stats import skew, kurtosis
from ucimlrepo import fetch_ucirepo

# fetch dataset
# https://archive.ics.uci.edu/ml/datasets/forest+fires
dforest = fetch_ucirepo(id=162)["data"]["original"]

# Select the variables of choice
variables = ["FFMC", "DMC", "DC", "ISI", "temp", "RH", "wind", "rain"]

# Skewness and kurtosis for each variable
skewness = dforest[variables].skew()
kurt = dforest[variables].kurtosis()

# Print the results and classify the kurtosis
# [E38] pandas returns the bias-corrected EXCESS kurtosis (normal = 0)
for var in skewness.index:
    print(f"{var} Skewness: {skewness[var]:.2f}")
    if kurt[var] > 0:
        print(f"{var} Kurtosis: {kurt[var]:.2f} (Leptokurtic)")
    elif kurt[var] < 0:
        print(f"{var} Kurtosis: {kurt[var]:.2f} (Platykurtic)")
    else:
        print(f"{var} Kurtosis: {kurt[var]:.2f} (Mesokurtic)")
```

```python
# Calculate Correlation Coefficients: Chi-square, Cramér's V and Point Biserial
# for the categorical variables of the Mammographic dataset
import pandas as pd
import numpy as np
from scipy import stats
from ucimlrepo import fetch_ucirepo
from scipy.stats import pointbiserialr

# fetch dataset
# https://archive.ics.uci.edu/ml/datasets/Mammographic+Mass
dmammo = fetch_ucirepo(id=161)["data"]["original"]
cols = ["BI-RADS", "Age", "Shape", "Margin", "Density", "Severity"]
dmammo = dmammo[cols]
dmammo.dropna(inplace=True) # Remove rows with missing values

# Chi-square and Cramér's V
cvars = ["Shape", "Margin", "Density", "Severity"] # Categorical variables
chis = pd.DataFrame()
phi = pd.DataFrame()
for var1 in cvars:
    for var2 in cvars:
        if var1 != var2:
            chi2, pval, dof, exp_freq = stats.chi2_contingency(pd.crosstab(dmammo[var1], dmammo[var2]))
            chis.loc[var1, var2] = chi2
            phi.loc[var1, var2] = np.sqrt(chi2 / (dmammo.shape[0] * (min(exp_freq.shape) - 1)))
        else: 
            chis.loc[var1, var2] = np.nan
            phi.loc[var1, var2] = np.nan
            
print("\n**Mammographic Dataset: Chi-Square, Cramér's V, PBCC**\n")
print("Chi-square statistic (chi^2)\n", chis)
print("\nCramér's V (phi)\n", phi)

# Point Biserial Correlation between Age and Severity
PBCC, pval = pointbiserialr(dmammo["Severity"], dmammo["Age"])
print(f"\nPBCC between Age and Severity: {PBCC:.2f}")
```

```python
# Scatter plots between pairs of variables from the Forest Fires dataset
import numpy as np
import scipy.stats as stats
import pandas as pd
from ucimlrepo import fetch_ucirepo
import matplotlib.pyplot as plt

# fetch dataset
# https://archive.ics.uci.edu/ml/datasets/forest+fires
dforest = fetch_ucirepo(id=162)["data"]["original"]

# Extract the relevant columns
cols = ["DMC", "DC", "FFMC", "ISI", "temp", "RH", "wind", "area", "rain"]
dforest = dforest[cols]

# Define the desired scatterplots
scatterplots = [("DMC", "DC"), ("FFMC", "ISI"), ("temp", "RH"), 
("temp", "wind"), ("wind", "area"), ("rain", "DC")]

# Generate the scatter plots
fig, axs = plt.subplots(nrows=3, ncols=2, figsize=(12, 14))
plt.subplots_adjust(wspace=0.4, hspace=0.8)

# Flatten the axs array for easier indexing
axs = axs.flatten()

# Function to plot linear regression line
def plot_regression_line(x, y, ax):
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    ax.plot(x, intercept + slope * x, "b-", label="Linear Regression")
    ax.legend()

for idx, (x, y) in enumerate(scatterplots):
    axs[idx].scatter(dforest[x], dforest[y], alpha=0.5)
    axs[idx].set_xlabel(x)
    axs[idx].set_ylabel(y)
    axs[idx].set_title(f'PCC: {dforest[x].corr(dforest[y], method="pearson"):.2f}', fontsize=16)
    plot_regression_line(dforest[x], dforest[y], axs[idx])

    # Set x & y-axis label font size
    axs[idx].set_xlabel(axs[idx].get_xlabel(), fontsize=16)
    axs[idx].set_ylabel(axs[idx].get_ylabel(), fontsize=16)
    
    # Set x & y-axis tick labels font size
    axs[idx].tick_params(labelsize=14)

# Save & Show the plot
plt.savefig("Figure_3_9_Scatterplot_Regression.svg", format="svg", dpi=1500)
plt.show()
```
